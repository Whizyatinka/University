### **Пункт №2. Обработка исключений в цикле**

Напишите программу, которая запрашивает у пользователя числа и добавляет их в список.
- Если пользователь вводит некорректные данные, программа должна выводить сообщение об ошибке и продолжать работу.
- Для выхода из программы пользователь должен ввести "стоп".

**ОБЯЗАТЕЛЬНО:**
- Покажите процесс ввода минимум 7 значений (включая ошибочные)
- Выведите финальный список и статистику (сколько чисел введено успешно, сколько ошибок)



```
# Пример:
# Введите число: 10
# Введите число: текст (Ошибка: это не число)
# Введите число: стоп
# Итоговый список: [10]
```



````python
numbers_list = []
correct = 0
incorrect = 0

while True:
    user_input = input('Введите число: ')

    if user_input.lower() == 'стоп':
        break

    try:
        value = float(user_input)
        numbers_list.append(value)
        correct += 1

    except ValueError:
        print(f"Неверный ввод: '{user_input}' не является числом")
        incorrect += 1

    print()

print(f"Полученные числа: {numbers_list}")
print(f"Успешных вводов: {correct}")
print(f"Ошибочных вводов: {incorrect}")
````

### **Пункт №3**

Создайте функцию `get_list_element(lst, index)`, которая безопасно извлекает элемент списка по индексу с использованием блоков `try`, `except`, `else` и `finally`.

Функция должна:
- Обрабатывать выход за пределы списка
- Обрабатывать некорректный тип индекса
- Выводить сообщение о попытке доступа в блоке `finally`

**ОБЯЗАТЕЛЬНО:**
- Выведите результаты для минимум 4 различных случаев (успешный доступ, выход за границы, отрицательный индекс, нечисловой индекс)
- Покажите, что блок `finally` выполняется всегда

```python
# Пример использования:
my_list = [1, 2, 3, 4, 5]
get_list_element(my_list, 2)   # Вывод: "Элемент: 3", "Попытка доступа завершена"
get_list_element(my_list, 10)  # Вывод: "Индекс вне диапазона", "Попытка доступа завершена"
```

````python
def get_list_element(collection, position):
    print(f"\nСписок {collection}, индекс {position}")
    try:
        if isinstance(position, str):
            position = int(position)
        item = collection[position]
    except ValueError:
        print(f"Некорректный индекс '{position}' - должно быть целое число")
    except IndexError:
        if position >= 0:
            print(f"Индекс {position} превышает размер списка")
        else:
            print(f"Индекс {position} выходит за границу списка")
    except TypeError as error:
        print(f"Неверный тип данных: {error}")
    else:
        print(f"Найден элемент {item} по индексу {position}")
        return item
    finally:
        print('Операция завершена')

data = [1, 2, 3, 4, 5]
print("\n1. Корректный доступ:")
first_result = get_list_element(data, 2)
print("\n2. Выход за пределы списка")
second_result = get_list_element(data, 10)
print("\n3. Отрицательный индекс")
third_result = get_list_element(data, -10)
print("\n4. Нецелочисленный индекс")
fourth_result = get_list_element(data, 1.234)
````

````python
def calculate_file_metrics(file_path):
    try:
        with open(file_path, 'r', encoding='utf-8') as file:
            text_content = file.read()

        line_count = text_content.splitlines()
        word_count = text_content.split()

        metrics = {
            'общее количество строк': len(line_count),
            'общее количество слов': len(word_count),
            'символов с пробелами': len(text_content),
            'символов без пробелов': len(text_content.replace(' ', ''))
        }

        return metrics

    except FileNotFoundError:
        print(f"Файл '{file_path}' не существует!")
        return None

sample_text = """Первая строка текста
Вторая строка с большим количеством слов
Третья строка
Четвертая строка для теста
Пятая и последняя строка"""

with open('test_file.txt', 'w', encoding='utf-8') as file:
    file.write(sample_text)

with open('test_file.txt', 'r', encoding='utf-8') as file:
    print(file.read())

file_metrics = calculate_file_metrics('test_file.txt')
if file_metrics:
    for metric_name, metric_value in file_metrics.items():
        print(f"{metric_name}: {metric_value}")
````

### **Пункт №9. Объединение файлов**

Создайте функцию `merge_files(file_list, output_file)`, которая объединяет содержимое нескольких текстовых файлов в один, добавляя после каждого разделитель.

**ОБЯЗАТЕЛЬНО:**
- Создайте минимум 3 файла с разным содержимым
- Выведите содержимое КАЖДОГО исходного файла
- Выполните объединение
- Выведите содержимое результирующего файла

````python
def combine_files(input_files, merged_file):
    try:
        with open(merged_file, 'w', encoding='utf-8') as output:
            for file_index, current_file in enumerate(input_files):
                with open(current_file, 'r', encoding='utf-8') as input_data:
                    output.write(input_data.read())

                if file_index < len(input_files) - 1:
                    output.write("\n\n\n")
        print(f"Результирующий файл создан: '{merged_file}'")
        return True
    except Exception as error:
        print(f"Произошла ошибка: {error}")
        return False

content_1 = """Исходный документ 1
Содержимое первой строки
Информация второй строки
Данные третьей строки"""

content_2 = """Дополнительный документ 2
Четвертая текстовая строка
Пятая информационная строка
Шестая завершающая строка"""

content_3 = """Финальный документ 3
Седьмая аналитическая строка
Восьмая итоговая строка
Девятая заключительная строка"""

with open('document1.txt', 'w', encoding='utf-8') as file:
    file.write(content_1)

with open('document2.txt', 'w', encoding='utf-8') as file:
    file.write(content_2)

with open('document3.txt', 'w', encoding='utf-8') as file:
    file.write(content_3)

documents = ['document1.txt', 'document2.txt', 'document3.txt']

for doc_name in documents:
    print(f"\nСодержимое файла {doc_name}:")
    with open(doc_name, 'r', encoding='utf-8') as file:
        print(file.read())

combine_files(documents, 'combined_document.txt')

with open('combined_document.txt', 'r', encoding='utf-8') as file:
    print("\nИтоговый объединенный документ:")
    print(file.read())
````
